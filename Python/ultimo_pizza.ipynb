```python
# Hemos puesto "import random" porque necesitamos la librería random para hacer cosas al azar.
# Queremos mezclar la lista de clientes y elegir qué pizza pide cada uno. Sin ella, el juego sería siempre igual.
import random
# Hemos puesto "import time" porque necesitamos la librería time para dos cosas:
# medir los 50 segundos que tiene el jugador y hacer la pausa de un segundo en la cuenta atrás del horno.
import time

from IPython.display import display, HTML, clear_output  # Eva

# Hemos puesto "def tarjeta" porque necesitamos dibujar el mismo recuadro muchas veces
# (bienvenida, cliente, resultados...). Con una función lo escribimos una sola vez y lo reutilizamos.
# Los parámetros son: titulo y contenido (lo que cambia en cada tarjeta) y color,
# que tiene un valor por defecto para no tener que indicarlo siempre.
def tarjeta(titulo, contenido, color="#8d3224"):  # Eva
    display(HTML(f'''<div style="max-width:740px;margin:12px auto;padding:20px 24px;border:1px solid #ead6c0;border-radius:18px;background:#fffaf3;color:#39281d;font-family:Arial,sans-serif;box-shadow:0 4px 14px #00000010">
    <div style="font-size:23px;font-weight:800;color:{color};margin-bottom:10px">{titulo}</div>
    <div style="font-size:16px;line-height:1.7">{contenido}</div></div>'''))

# "iconos" como diccionario porque necesitamos relacionar cada ingrediente con su emoji.
# Es un diccionario (y no una lista) porque así buscamos el emoji directamente por el nombre del ingrediente.
iconos = {"masa": "🫓", "tomate": "🍅", "queso": "🧀", "mix 4 quesos": "🧀", "jamón": "🥓", "pepperoni": "🍕"}  # Eva

# "clear_output" porque necesitamos que, si se ejecuta el juego otra vez,
# no se acumule lo de la partida anterior. Con wait=True espera a tener algo nuevo antes de borrar.
clear_output(wait=True)  # Eva

# La lista "clientes" para saber el género de cada uno de los 5 clientes.
clientes = ["Mujer", "Mujer", "Mujer", "Hombre", "Hombre"]
# "random.shuffle" para que el orden de los clientes cambie en cada partida.
# Si no, el cliente 1 sería siempre una mujer.
random.shuffle(clientes)

# "dinero = 50" porque necesitamos una variable que guarde el saldo del jugador.
# Empieza en 50 porque es el dinero inicial del juego. Luego sumamos y restamos según aciertos y fallos.
dinero = 50
# "clientes_satisfechos = 0" porque necesitamos contar cuántos clientes se han atendido bien.
# Empieza en 0 porque al principio nadie está satisfecho. Al final se compara con 5 para saber si se gana.
clientes_satisfechos = 0

# "def ingrediente(numero)" porque necesitamos convertir el número que escribe el jugador, cada número es un ingrediente.
# Hemos puesto "def ingrediente(numero)" porque necesitamos convertir el número que escribe el jugador
# (del 1 al 6) en el nombre del ingrediente. Lo hacemos función porque la usamos varias veces.
def ingrediente(numero):
    # Hemos creado esta lista porque necesitamos tener los 6 ingredientes ordenados.
    # Cada uno queda en una posición de la lista, empezando por la 0, masa.
    ingredientes = ["masa", "tomate", "queso", "mix 4 quesos", "jamón", "pepperoni"]
    # Hemos puesto "return" porque necesitamos que la función nos devuelva el ingrediente para poder usarlo fuera
    # Restamos 1 porque el jugador escribe del 1 al 6, las listas empiezan en la posición 0.
    return ingredientes[numero - 1]

# El diccionario "recetas" es porque necesitamos saber qué ingredientes lleva cada pizza.
# Cada pizza (clave) tiene su lista de ingredientes (valor). 
# Lo usamos al final del turno para comprobar si la pizza que ha hecho el jugador coincide con la que pidió el cliente.
recetas = {
    "margarita": ["masa", "tomate", "queso"],
    "4 quesos": ["masa", "tomate", "mix 4 quesos"],
    "jamón y queso": ["masa", "tomate", "queso", "jamón"],
    "pepperoni": ["masa", "tomate", "queso", "pepperoni"]
}

display(HTML('''<div style="max-width:740px;margin:14px auto 20px;padding:30px 22px;border-radius:20px;background:linear-gradient(135deg,#842b24,#c65d31);text-align:center;color:white;font-family:Arial,sans-serif;box-shadow:0 7px 20px #00000025">
<div style="font-size:36px;font-weight:900">🍕 PIZZA PATRICCINI 🍕</div>
<div style="font-size:18px;margin:12px 0"><i>¡Ponte el delantal y enciende el horno!</i></div>
<div style="font-size:16px;font-weight:bold">⭐ NIVEL 1 · CINCO CLIENTES ⭐</div></div>'''))  # Eva
# Hemos puesto esta tarjeta porque necesitamos explicar al jugador las reglas antes de empezar.
# Lleva "f" porque mete la variable {dinero}. Las etiquetas <b> (negrita) y <br> (salto de línea) son HTML.
tarjeta("📜 BIENVENIDO A PIZZA PATRICCINI", f"Atiende correctamente a los <b>5 clientes</b>.<br>💰 Empiezas con <b>{dinero} €</b>.<br>Cada pizza correcta suma <b>20 €</b> y cada incorrecta resta <b>10 €</b>.<br>⏳ Tienes <b>50 segundos</b> por cliente. Si se agota el tiempo, pierdes <b>5 €</b> y tienes una segunda oportunidad.")  # Eva

# Hemos puesto un "for" porque necesitamos repetir el mismo proceso (pedido, ingredientes, comprobación)
# para cada uno de los 5 clientes sin copiar el código 5 veces.
# range(1, 6) genera una secuencia de números: 1, 2, 3, 4 y 5; el 6 no se incluye. 
# En cada vuelta, la variable "cliente" vale el número del cliente que estamos atendiendo.
for cliente in range(1, 6):

    # Hemos puesto "random.choice" porque necesitamos que cada cliente pida una pizza al azar.
    # list(recetas.keys()) convierte los nombres de las pizzas en una lista, que es lo que random.choice necesita para poder elegir una.
    pizza_pedida = random.choice(list(recetas.keys()))

    # Hemos creado "genero" porque necesitamos saber si este cliente es mujer u hombre.
    # Usamos "cliente - 1" porque el cliente 1 está en la posición 0 de la lista "clientes".
    genero = clientes[cliente - 1]  # Eva
    # Hemos puesto esta tarjeta porque necesitamos presentar al cliente, mostrar lo que pide y recordar al jugador su saldo y sus aciertos.
    # El "'Una' if genero == 'Mujer' else 'Un'" sirve para decir "Una mujer" o "Un hombre" según el género.
    # .lower() pone el género en minúsculas y .capitalize() pone en mayúscula solo la primera letra de la pizza.
    tarjeta(f"👩‍🍳 CLIENTE {cliente} DE 5", f"{'Una' if genero == 'Mujer' else 'Un'} <b>{genero.lower()}</b> se acerca al mostrador y coge el menú.<br>🍕 <i>Me gustaría una pizza {pizza_pedida.capitalize()}.</i><br>💰 Saldo: <b>{dinero} €</b> &nbsp;·&nbsp; ⭐ Clientes satisfechos: <b>{clientes_satisfechos}/5</b>")  # Eva

    # Hemos creado "fallos_tiempo = 0" porque necesitamos contar cuántas veces se acaba el tiempo con este cliente.
    # Está dentro del for para que vuelva a 0 con cada cliente nuevo.
    fallos_tiempo = 0
    # Esta variable la creamos pero luego no se usa en ningún sitio, así que se podría borrar
    # sin que cambie nada en el juego.
    pizza_correcta = False

    # Hemos creado "opciones" porque necesitamos el texto del menú de ingredientes.
    # El "for n in range(1, 7)" de dentro recorre los números del 1 al 6 y para cada uno escribe "número + emoji + nombre". " | ".join(...) los une con una barra entre medias.
    # .title() pone en mayúscula la primera letra de cada palabra.
    opciones = " &nbsp; | &nbsp; ".join(f"<b>{n}</b> {iconos[ingrediente(n)]} {ingrediente(n).title()}" for n in range(1, 7))  # Eva
    # Hemos puesto esta tarjeta porque necesitamos enseñar al jugador qué ingredientes hay y qué número
    # tiene cada uno, y recordarle que con 0 termina la pizza.
    tarjeta("🧺 INGREDIENTES", opciones + "<br><i>Pulsa 0 cuando hayas terminado.</i>", "#985b15")  # Eva

    # Hemos creado "inicio = time.time()" porque necesitamos guardar la hora exacta a la que empieza el turno.
    # Más adelante restamos esta hora a la hora actual para saber cuántos segundos han pasado.
    inicio = time.time()

    # Hemos creado "pizza = []" porque necesitamos una lista donde ir guardando los ingredientes
    # que elige el jugador. Empieza vacía porque todavía no ha puesto nada. Los ingredientes los vamos añadiendo a medida q jugamos
    pizza = []

    # Hemos puesto "while True" porque necesitamos que el jugador pueda elegir ingredientes las veces que quiera, ya que no sabemos cuántos va a poner. Es un bucle.
    # lo cortamos nosotros con "break" cuando termina o se acaba el tiempo.
    while True:

    tiempo_restante = 50 - int(time.time() - inicio)

    print("\nTiempo restante:", tiempo_restante, "segundos")

    numero = int(input("Elige tus ingredientes, pulsa 0 cuando hayas terminado: "))

    if numero == 0:
        break
   

        # Este "if" es porque necesitamos saber si se ha pasado el límite de 50 segundos.
        # Se comprueba justo después del input porque el programa no puede interrumpir al jugador mientras escribe: solo se da cuenta del tiempo cuando responde.
        if time.time() - inicio >= 50:

            # "fallos_tiempo += 1" para apuntar que se ha acabado el tiempo.
            # Es lo mismo que fallos_tiempo = fallos_tiempo + 1. Luego lo usamos para saber si es el primer fallo (segunda oportunidad) o el segundo (el cliente se va).
            fallos_tiempo += 1
            # Hemos puesto "dinero -= 5" porque necesitamos aplicar la penalización de 5 € por tardar.
            dinero -= 5

            # Hemos puesto esta tarjeta porque necesitamos avisar al jugador de que se ha acabado el tiempo y de cuánto dinero le queda.
            tarjeta("⏰ ¡SE ACABÓ EL TIEMPO!", f"Pierdes <b>5 €</b>. Saldo actual: <b>{dinero} €</b>.", "#b04c25")  # Eva

            # Este "if" es porque necesitamos distinguir el primer fallo del segundo.
            # Solo en el primero damos otra oportunidad.
            if fallos_tiempo == 1:
                tarjeta("💛 SEGUNDA OPORTUNIDAD", "El cliente te da una segunda oportunidad.")  # Eva

                # Hemos puesto "inicio = time.time()" otra vez porque necesitamos reiniciar el reloj para que el jugador vuelva a tener 50 segundos completos.
                inicio = time.time()
                # Hemos puesto "pizza = []" porque necesitamos vaciar la pizza y que empiece de cero en el segundo intento.
                pizza = []

                # Hemos puesto "continue" porque necesitamos volver al principio del while (mostrar el tiempo y pedir ingrediente) sin ejecutar el resto del código de esta vuelta.
                continue

            # "else" porque necesitamos decir qué pasa en el resto de casos:
            # si no es el primer fallo, es el segundo y no hay más oportunidades.
            else:
                tarjeta("🚶 EL CLIENTE SE MARCHA", "El cliente se ha cansado y se va.", "#b04c25")  # Eva
                # Hemos vaciado la pizza porque el cliente se va sin ella.
                pizza = []
                # Hemos puesto "break" porque necesitamos salir del while: ya no se pueden elegir más ingredientes para este cliente.
                break

        # Este "if" porque necesitamos una forma de que el jugador diga que ha terminado:escribiendo 0. 
        # El break termina el bucle while: dejamos de pedir ingredientes y seguimos con el código que viene después.
          
        # Este "if" es porque necesitamos evitar que el jugador escriba un número que no corresponde a ningún ingrediente. 
        # El "or" significa que basta con que se cumpla una de las dos condiciones: menor que 1 o mayor que 6. 
        # Sin esto, ingrediente() daría error con números como 9.
        if numero < 1 or numero > 6:
            print("⚠️ Ese ingrediente no existe. Elige un número del 1 al 6.")  # Eva (emoji)
            # Hemos puesto "continue" porque necesitamos que el jugador pueda elegir de nuevo sin perder lo que lleva en la pizza ni reiniciar el tiempo.
            continue

        # Hemos puesto este print porque necesitamos confirmar al jugador qué ingrediente acaba de añadir.
        # Usamos ingrediente(numero) nos da el nombre (por ejemplo "queso"), iconos[...] busca su emoji, y .title() pone la primera letra en mayúscula para que quede "Queso".
        print(
    f"✅ Has añadido: {iconos[ingrediente(numero)]} "
    f"{ingrediente(numero).title()} | "
    f"Tu pizza: {', '.join(pizza)}"
)
        pizza.append(ingrediente(numero))

    # Hemos puesto este print porque necesitamos enseñar al jugador con qué ingredientes ha hecho la pizza antes de comprobar si es correcta.
    print("Tu pizza lleva:", pizza)
    # Este "if" es porque necesitamos saltarnos la comprobación cuando el cliente se ha ido.
    # Si fallos_tiempo es 2, el cliente se ha ido sin pizza. 
    # Hacemos continue para pasar al siguiente cliente y no comprobar una pizza vacía, que daría "incorrecta" y nos quitaría 10 € extra, además de los 5 € que ya hemos perdido por el tiempo.
    if fallos_tiempo == 2:
        continue
    # Este "if" es porque necesitamos comprobar si la pizza del jugador es la que pidió el cliente.
    # sorted() ordena las dos listas antes de compararlas, así no importa en qué orden se hayan puesto los ingredientes.
    # Pero sí importa que no falte ninguno, que no sobre ninguno y que no se repita.
    # recetas[pizza_pedida] busca en el diccionario la lista correcta de esa pizza.
    if sorted(pizza) == sorted(recetas[pizza_pedida]):
 
        # Hemos puesto esta tarjeta porque necesitamos avisar al jugador de que la pizza está bien.
        tarjeta("🔥 ¡PIZZA CORRECTA!", "<i>¡Qué buena pinta!</i> La pizza está en el horno...", "#23734e")  # Eva
 
        # Hemos puesto este "for" porque necesitamos una cuenta atrás que simule el horneado.
        # range(10, 0, -1) cuenta hacia atrás de 10 a 1: el 10 es el inicio, el 0 es el final (no se incluye) y el -1 es el paso, es decir, va restando de uno en uno.
        for segundos in range(3, 0, -1):
            print("🔥 Horno:", segundos, "segundos restantes")  # Eva (emoji)
            # Hemos puesto "time.sleep(1)" porque necesitamos que pase un segundo real entre cada número.
            # Sin esta pausa, los 10 números saldrían de golpe y no parecería un horno.
            time.sleep(1)
 
        # Hemos puesto "dinero += 20" porque necesitamos dar el premio por la pizza correcta.
        dinero += 20
        # Hemos puesto "clientes_satisfechos += 1" porque necesitamos apuntar que este cliente está contento.
        # Al final, este contador decide si se supera el nivel.
        clientes_satisfechos += 1
 
        # Hemos puesto esta tarjeta porque necesitamos informar del premio y del nuevo saldo.
        tarjeta("🎉 ¡CLIENTE SATISFECHO!", f"Has ganado <b>20 €</b>. Saldo actual: <b>{dinero} €</b>.", "#23734e")  # Eva
 
    # "else" para decir qué pasa cuando la pizza NO es la correcta.
    # Se ejecuta cuando la comparación del "if" de arriba da falso.
    else:
 
        # Hemos puesto "dinero -= 10" porque necesitamos aplicar la penalización por pizza incorrecta.
        dinero -= 10
 
        # Hemos puesto esta tarjeta porque necesitamos avisar de que la pizza estaba mal y mostrar cuánto dinero queda.
        tarjeta("❌ PIZZA INCORRECTA", f"Pierdes <b>10 €</b>. Saldo actual: <b>{dinero} €</b>.", "#a63b34")  # Eva
 
# Hemos puesto esta última tarjeta fuera del "for" (sin sangría) porque necesitamos que se muestre una sola vez, cuando ya se han atendido los 5 clientes.
# Dentro hay dos "A if condición else B": el primero elige entre el mensaje de victoria y el de derrota, y el segundo elige el color (verde si se gana, rojo si se pierde). Se gana solo si los 5 están satisfechos.
# El + sirve para pegar el marcador (clientes y dinero) con ese mensaje final en un solo texto.
tarjeta("🏁 FIN DEL NIVEL", f"⭐ Clientes satisfechos: <b>{clientes_satisfechos}/5</b><br>💰 Dinero final: <b>{dinero} €</b><br>" + ("🏆 <b>¡NIVEL SUPERADO!</b>" if clientes_satisfechos == 5 else "❌ <b>NIVEL NO SUPERADO</b>"), "#23734e" if clientes_satisfechos == 5 else "#a63b34")  # Eva
```
